```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv("dataset/Mall_Customers.csv")

# Display the first 5 rows
df.head()
```

```python
# Check the size of the dataset
print("Dataset shape:", df.shape)

# Check column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Check duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())

# Check unique values in Gender
print("\nUnique Gender values:")
print(df["Gender"].unique())

# Check the range of Age
print("\nAge range:")
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())

# Check the range of Annual Income
print("\nAnnual Income range:")
print("Minimum:", df["Annual Income (k$)"].min())
print("Maximum:", df["Annual Income (k$)"].max())

# Check the range of Spending Score
print("\nSpending Score range:")
print("Minimum:", df["Spending Score (1-100)"].min())
print("Maximum:", df["Spending Score (1-100)"].max())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Standardize column names
df.columns = [
    "customer_id",
    "gender",
    "age",
    "annual_income",
    "spending_score"
]

# Display the new column names
print("Updated column names:")
print(df.columns.tolist())
```

```python
# Check the data types after cleaning
print("Data types:")
print(df.dtypes)
```

```python
# Final data quality check

print("===== FINAL DATA QUALITY CHECK =====")

# Missing values
print("\nMissing values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())

# Check valid ranges
print("\nAge range:", df["age"].min(), "to", df["age"].max())
print("Annual income range:", df["annual_income"].min(), "to", df["annual_income"].max())
print("Spending score range:", df["spending_score"].min(), "to", df["spending_score"].max())

# Check gender values
print("\nGender values:")
print(df["gender"].unique())

# Check data types
print("\nData types:")
print(df.dtypes)
```

```python
# Save the cleaned dataset
output_path = "dataset/cleaned_mall_customers.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_path)
```
